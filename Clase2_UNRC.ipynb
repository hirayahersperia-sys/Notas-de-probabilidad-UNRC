```python
from itertools import product

S = list(product(range(1,7), repeat=2))
print("Espacio Muestral:", S)

A = [x for x in S if x[0] + x[1] == 7]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```

```python
from itertools import product

S = list(product(["A","S"], repeat=3))
print("Espacio Muestral:", S)

A = [x for x in S if x.count("A") == 2]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```

**Ejercicio 1**

Al lanzar dos monedas al aire, calcula la probabilidad de que ambas caigan en la misma cara (ambas águila o ambas sol)

```python
from itertools import product

S = list(product(["A","S"], repeat=2))
print("Espacio Muestral:", S)

A = [x for x in S if x[0] == x[1]]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```

**Ejercicio 2**

Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

```python
from itertools import product

S = list(product(range(1,7), repeat=2))
print("Espacio Muestral:", S)

A = [x for x in S if x[0] + x[1] == 5]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```
